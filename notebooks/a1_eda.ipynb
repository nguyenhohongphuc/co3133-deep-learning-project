```python
from pathlib import Path
import json

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from torchvision.datasets import FashionMNIST

# Hỗ trợ chạy notebook từ root repo hoặc từ thư mục notebooks/.
PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "configs").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data"
SPLIT_FILE = PROJECT_ROOT / "results/a1/splits/split_seed42.json"
FIGURE_DIR = PROJECT_ROOT / "results/a1/figures"
TABLE_DIR = PROJECT_ROOT / "results/a1/tables"

FIGURE_DIR.mkdir(parents=True, exist_ok=True)
TABLE_DIR.mkdir(parents=True, exist_ok=True)

with SPLIT_FILE.open(encoding="utf-8") as file:
    split = json.load(file)

dataset = FashionMNIST(root=DATA_DIR, train=True, download=False)

class_names = dataset.classes
labels = np.asarray(dataset.targets)

train_idx = np.asarray(split["train_idx"])
val_idx = np.asarray(split["val_idx"])

print("Classes:", class_names)
print("Raw train image tensor:", dataset.data.shape)
print("Raw pixel dtype:", dataset.data.dtype)
print("Seed:", split["seed"])
print("Split counts:", split["counts"])
```
